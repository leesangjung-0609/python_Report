```python
import numpy as np

class OmokEnvironment:
    def __init__(self):
        self.board = np.zeros((15, 15), dtype=int)
        self.current_turn = 1 # 1: 흑돌, 2: 백돌

    def reset(self):
        self.board = np.zeros((15, 15), dtype=int)
        self.current_turn = 1
        return self.board.copy()

    def step(self, x, y):
        # 1. 반칙패 처리
        if self.board[y][x] != 0:
            return self.board.copy(), -1, True, "반칙패"
            
        # 2. 착수
        self.board[y][x] = self.current_turn
        
        # 3. 승리 체크
        if self.check_win(x, y, self.current_turn):
            return self.board.copy(), 1, True, "승리"
            
        # 4. 무승부 체크
        if np.all(self.board != 0):
            return self.board.copy(), 0, True, "무승부"
            
        # 5. 턴 교대
        self.current_turn = 3 - self.current_turn
        return self.board.copy(), 0, False, "진행중"

    def check_win(self, x, y, stone):
        directions = [(1, 0), (0, 1), (1, 1), (1, -1)]
        for dx, dy in directions:
            count = 1
            nx, ny = x + dx, y + dy
            while 0 <= nx < 15 and 0 <= ny < 15 and self.board[ny][nx] == stone:
                count += 1
                nx += dx
                ny += dy
            nx, ny = x - dx, y - dy
            while 0 <= nx < 15 and 0 <= ny < 15 and self.board[ny][nx] == stone:
                count += 1
                nx -= dx
                ny -= dy
            if count >= 5:
                return True
        return False
```
